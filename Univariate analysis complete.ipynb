```python
plot_col=["Age","PlayTimeHours","InGamePurchases","SessionsPerWeek","AvgSessionDurationMinutes","PlayerLevel","AchievementsUnlocked"]
for col in plot_col:
    plt.figure()
    sns.kdeplot(df[col])
    plt.show()
```

```python
plot_col=["Age","PlayTimeHours","InGamePurchases","SessionsPerWeek","AvgSessionDurationMinutes","PlayerLevel","AchievementsUnlocked"]
for col in plot_col:
    skewness=df[col].skew()
    plt.figure()
    sns.kdeplot(df[col],fill=True)
    plt.title(f"{col} | skewness: {skewness:.3f}")
    plt.tight_layout()
    plt.show()
```

```python
plot_col=["Age","PlayTimeHours","InGamePurchases","SessionsPerWeek","AvgSessionDurationMinutes","PlayerLevel","AchievementsUnlocked"]
for col in plot_col:
    skewness=df[col].skew()
    kurtosis=df[col].kurt()
    plt.figure()
    sns.kdeplot(df[col],fill=True)
    plt.title(f"{col} | skewness:{skewness:.2f} | kurtosis: {kurtosis:.2f}")
    plt.tight_layout()
    plt.show()
```

```python
nd=["Age","PlayTimeHours","SessionsPerWeek","AvgSessionDurationMinutes","PlayerLevel","AchievementsUnlocked"]
for x in nd:
    plt.figure()
    plt.boxplot(df[x])
    plt.title(f"{x}")
    plt.tight_layout()
    plt.show()
```
