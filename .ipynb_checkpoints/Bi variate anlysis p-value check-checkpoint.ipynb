```python
from scipy import stats # for t-test, ANOVA, Chi-square test
```

### Finding p-value
Signal vs Noise score

```python
contingency_table=[[20,10], [25,15]]
chi2, p_val_chi, dof, expected=stats.chi2_contingency(contingency_table)
print(f'{p_val_chi:.4f}')
```

```python
# > 0.05 is noise 
# < 0.05 is signal
```
