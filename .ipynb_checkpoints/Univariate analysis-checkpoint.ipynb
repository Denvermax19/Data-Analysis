# Count of the Wine with its quality rate.

```python
wine_quality=df["quality"].value_counts()
plt.figure()
plt.bar(wine_quality.index,wine_quality)
plt.show()
```

## With improved features for creating bar chart

```python
wine_quality=df["quality"].value_counts()
plt.figure()
plt.bar(wine_quality.index,wine_quality)
plt.xlabel("Wine Quality")
plt.ylabel("Count")
plt.title("Wine quality metrics")
plt.show()
```

### With save feature

```python
wine_quality=df["quality"].value_counts()
plt.figure()
plt.bar(wine_quality.index,wine_quality)
plt.xlabel("Wine Quality")
plt.ylabel("Count")
plt.title("Wine quality metrics")
plt.savefig("wine quality chart", dpi=300)  # For saving the charts, dpi for quality of image
plt.show()
```

## Understanding Variance in the dataset

```python
# we are using Kernal Density plots
```

```python
for col in df.columns:
    plt.figure()
    sns.kdeplot(df[col], fill=True)
    plt.tight_layout()
    plt.show()
```

#### Skewness = Mean -- Mode

```python
for col in df.columns:
    skew_value=df[col].skew()
    plt.figure()
    sns.kdeplot(df[col], fill=True)
    plt.tight_layout()
    plt.title(f"KDE plot {col} | skewness: {skew_value:.3f}")
    plt.show()
```
