```python
df=pd.read_csv(r"C:\Users\lette\Data Analytics\Data Visualization\Zomato-data-.csv")
df.head()
```

```python
# in the 'rate' column the data is irregular format which is not good for data analysis

# so we make {4.1/5} as {4.5}
```

```python
def clean_rate(value):
    value=str(value).split('/')
    value=value[0]
    return float(value)
df['rate']=df['rate'].apply(clean_rate)
df.head()
```

```python
# few columns are not convenient to read, so we make them better readable
```

```python
df.rename(columns={'approx_cost(for two people)':'cost for 2 people','listed_in(type)':'restaurant type'},inplace=True)
```

```python
df.head()
```

```python
df.describe()
```

```python
#plt.figure() # optional for setting figure sizes if needed
sns.countplot(x=df['online_order'],color='skyblue')
plt.show()
```

```python
#plt.figure() # optional for setting figure sizes if needed
sns.countplot(x=df['book_table'],color="lightpink")
plt.show()
```

```python
#plt.figure() # optional for setting figure sizes if needed
sns.countplot(x=df['restaurant type'],color='lightgreen')
plt.show()
```

```python
# restaurants with highest Votes

top_restaurants=df['votes'].nlargest(5).index
selected_restaurants=df.loc[top_restaurants,['name','votes']]

sns.lineplot(x=selected_restaurants['name'],y=selected_restaurants['votes'],color='red',marker='o')
plt.show()
```

```python
# Restaurants Count with Given Votes Range
def restaurants_count(a,b):
    checking=df['votes'].between(a,b).sum()
    return(f'Number of restaurants: {checking}')

restaurants_count(0,0)
```

```python
# Votes based on Restaurant Type
type_grouped=df.groupby('restaurant type')['votes'].sum()
grouped_df=pd.DataFrame({'votes':type_grouped})
sns.lineplot(grouped_df,palette='hls',markers='o')
plt.show()
```

```python
# Checking rating Distribution

sns.kdeplot(df['rate'],fill=True,color='orange')
plt.show()
```
