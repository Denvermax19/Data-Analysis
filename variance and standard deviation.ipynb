```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Generate data
low_var = np.array([4,5,5,6,5,4,5,6,5])
high_var = np.array([1,9,2,10,3,11,4,12])
```

```python
# First figure: Variance explanation (low vs high)
plt.figure()
plt.scatter(range(len(low_var)), low_var)
plt.title("Low Variance Example")
plt.xlabel("Kids")
plt.ylabel("Candies")
plt.show()
```

```python
plt.figure()
plt.scatter(range(len(high_var)), high_var)
plt.title("High Variance Example")
plt.xlabel("Kids")
plt.ylabel("Candies")
plt.show()
```

```python
# Second figure: Standard Deviation visual with mean line
data = np.array([3,4,5,6,7,8,9])
mean = data.mean()

plt.figure()
plt.scatter(range(len(data)), data)
plt.axhline(mean)
plt.title("Standard Deviation: How Much Numbers Wiggle From the Mean")
plt.xlabel("Kids")
plt.ylabel("Candies")
plt.show()
```
