```python
import numpy as np
```

```python
low_var=np.array([4,5,5,6,5,4,5,6,5])
high_var=np.array([1,9,2,10,3,11,4,12])
```

```python
np.mean(low_var) # MEAN
```

```python
np.median(low_var) # MEDIAN
```

```python
np.ptp(low_var)  # RANGE
```

```python
np.var(low_var) # VARIANCE
```

```python
np.var(high_var) # VARIANCE FOR 2ND VARIABLE
```

```python
np.std(low_var) # STANDARD DEVIATION FOR 1ST VARIABLE
```

```python
np.std(high_var) # STANDARD DEVIATION FOR 2ND VARIABLE
```

Measuring Central Tendency Using Pandas

```python
import pandas as pd
```

```python
df=pd.DataFrame(low_var,columns=["chocolates"])
```

```python
df.mean()
```

```python
df.median()
```
